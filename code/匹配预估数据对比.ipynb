## 更新 预估数据对比 表

```python
# 将 Listing预估表（模板，15，30，60） 数据更新到 预估数据对比表

import pandas as pd
year = input('请输入年份：')
month = input('请输入月份：')
days = input('请输入日期：')
parms = f'{year}{month}{days}'
```

```python
# 读取 预估数据对比表
df = pd.read_excel('E:/sontu/shipping_recommendations/预估数据对比.xlsx')
# 读取 Listing预估表-模板
df_temp = pd.read_excel(f'E:/sontu/shipping_recommendations/src_data/Listing预估表-模板.xlsx')
# 读取 Listing预估表-15
df_15 = pd.read_excel(f'E:/sontu/shipping_recommendations/src_data/Listing预估表-{parms}-15.xlsx')
# 读取 Listing预估表-30
df_30 = pd.read_excel(f'E:/sontu/shipping_recommendations/src_data/Listing预估表-{parms}-30.xlsx')
# 读取 Listing预估表-60
df_60 = pd.read_excel(f'E:/sontu/shipping_recommendations/src_data/Listing预估表-{parms}-60.xlsx')
```

```python
# 构建各表的key，用于匹配
df['key'] = df['站点'].astype(str) + df['Listing'].astype(str) + df['月份'].astype(str)
df_temp['key'] = df_temp['站点'].astype(str) + df_temp['Listing'].astype(str) + df_temp['月份'].astype(str)
df_15['key'] = df_15['站点'].astype(str) + df_15['Listing'].astype(str) + df_15['月份'].astype(str)
df_30['key'] = df_30['站点'].astype(str) + df_30['Listing'].astype(str) + df_30['月份'].astype(str)
df_60['key'] = df_60['站点'].astype(str) + df_60['Listing'].astype(str) + df_60['月份'].astype(str)
# 删除不需要的列
df_temp = df_temp.drop(columns=['站点', 'Listing', '月份'])
df_15 = df_15.drop(columns=['站点', 'Listing', '月份'])
df_30 = df_30.drop(columns=['站点', 'Listing', '月份'])
df_60 = df_60.drop(columns=['站点', 'Listing', '月份'])
```

```python
yeara = year[2:]
# 匹配数据
df_merge = pd.merge(df, df_temp, on = ['key'], how = 'left')
df_merge = df_merge.rename(columns = {'Listing_月度预估销量':f'预估销量_{yeara}_{month}{days}'})

df_merge = pd.merge(df_merge, df_15, on = ['key'], how = 'left')
df_merge = df_merge.rename(columns = {'Listing_月度预估销量':f'预估销量_15_{yeara}_{month}{days}'})

df_merge = pd.merge(df_merge, df_30, on = ['key'], how = 'left')
df_merge = df_merge.rename(columns = {'Listing_月度预估销量':f'预估销量_30_{yeara}_{month}{days}'})

df_merge = pd.merge(df_merge, df_60, on = ['key'], how = 'left')
df_merge = df_merge.rename(columns = {'Listing_月度预估销量':f'预估销量_60_{yeara}_{month}{days}'})
```
